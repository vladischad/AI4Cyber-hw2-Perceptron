```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

class AvgPerceptron:
    def __init__(self, max_iter=40):
        self.max_iter = max_iter

        # initialize weights placeholders
        self.w = None
        self.b = 0.0
        self.avg_w = None
        self.avg_b = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape

        # initialize weights and bias
        self.w = np.zeros(n_features)
        self.b = 0.0

        sum_w = np.zeros(n_features)
        sum_b = 0.0
        total_steps = 0

        # process training
        for epoch in range(self.max_iter):
            for i in range(n_samples):
                # prediction
                activation = np.dot(X[i], self.w) + self.b

                # update weights and bias when incorrectly classified
                if y[i] * activation <= 0:
                    self.w += y[i] * X[i]
                    self.b += y[i]

                # hold information to calculate avg wieghts and bias
                sum_w += self.w
                sum_b += self.b
                total_steps += 1

        # calc final avg weights and bias based on sums
        self.avg_w = sum_w / total_steps
        self.avg_b = sum_b / total_steps

    def predict(self, X):
        return np.where((np.dot(X, self.avg_w) + self.avg_b) >= 0, 1, -1)
```
